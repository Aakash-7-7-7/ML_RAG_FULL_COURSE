```python
import sqlite3
import csv

conn = sqlite3.connect("agent_memory.db")
cursor = conn.cursor()

cursor.execute("""
    SELECT
        conversation_id,
        id AS message_id,
        role,
        content,
        extra_metadata,
        created_at
    FROM messages
    ORDER BY conversation_id, created_at
""")

rows = cursor.fetchall()

with open("all_conversations.csv", "w", newline="", encoding="utf-8-sig") as file:

    writer = csv.writer(file)

    writer.writerow([
        "conversation_id",
        "message_id",
        "role",
        "content",
        "extra_metadata",
        "created_at"
    ])

    writer.writerows(rows)

conn.close()

print(f"Done! Exported {len(rows)} messages.")
print("File: all_conversations.csv")
```
