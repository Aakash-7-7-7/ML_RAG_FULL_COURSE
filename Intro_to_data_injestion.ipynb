## TExt-Loader ---- SingleFIle

```python
from langchain_community.document_loaders import UnstructuredWordDocumentLoader

loader=UnstructuredWordDocumentLoader("docs/Evolution_of_Video_Games_and_AI.docx")
documents=loader.load()
print(documents)

print(f"Content: {documents[0].page_content[:500]}")


print(f"Metadata: {documents[0].metadata}")
```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader=DirectoryLoader(
    "docs",
    glob="**/*.docx",
    loader_cls=UnstructuredWordDocumentLoader,
    show_progress=True
)



doc=dir_loader.load()

print(f"Loaded {len(doc)} documents")

for i ,doc in enumerate(doc):
    print(f"\nDocument {i+1}:")
    print(f" Source: {doc.metadata['source']}")
    print(f" Length: {len(doc.page_content)}")
```
