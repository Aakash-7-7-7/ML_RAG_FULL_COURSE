```python
from typing import TypedDict
from langgraph.graph import StateGraph,START,END

class State(TypedDict):
    name:str
    greeting:str


def create_greeting(state:State):
    return{
        "greeting": f"Hello {state['name']}!"
    }

def final_response(state:State):
    print(state["greeting"])


builder=StateGraph(State)

builder.add_node("create_greeting",create_greeting)
builder.add_node("final_response",final_response)

builder.add_edge(START,'create_greeting')
builder.add_edge("create_greeting","final_response")
builder.add_edge("final_response",END)

app=builder.compile()

app.invoke({
    "name":"AAKASH",
    "greeting":""
})
```

```python
from dataclasses import dataclass
from langgraph.graph import StateGraph, START, END


# 1. Define State using dataclass
@dataclass
class State:
    name: str
    age: int
    message: str = ""


# 2. First node
def node1(state: State):
    return {
        "message": f"Hello {state.name}, you are {state.age} years old."
    }


# 3. Second node
def node2(state: State):
    print(state.message)


# 4. Create graph
builder = StateGraph(State)

builder.add_node("node1", node1)
builder.add_node("node2", node2)

builder.add_edge(START, "node1")
builder.add_edge("node1", "node2")
builder.add_edge("node2", END)


# 5. Compile
app = builder.compile()


# 6. Run
app.invoke({
    "name": "Aakash",
    "age": 22,
    "message": ""
})
```
