## Core Expression Language (Tables, Columns, Types)

```python
from sqlalchemy import MetaData , String , Integer , Column , Table , ForeignKey , Boolean ,create_engine , text ,select , insert

engine=create_engine("sqlite:///Core.db" , echo=True)

metadata=MetaData()

users=Table(
    "users",metadata,
    Column("id", Integer , primary_key=True),
    Column("name", String , nullable= False),
    Column("email", String , nullable=False, unique=True)
)

orders=Table(
    "orders",metadata,
    Column("id" , Integer , primary_key=True),
    Column("user_id", Integer , ForeignKey("users.id")),
    Column("total", Integer)
)

metadata.create_all(engine)
conn=engine.connect()
```

```python
insert_users=users.insert().values([
    {"id": 1, "name":"John" , "email":"John@gmail.com" },
    {"id": 2, "name": "Tyler", "email": "Tyler@gmail.com"},
    {"id": 3, "name": "Tyrell", "email": "Tyrell@gmail.com"},
    {"id": 4, "name": "Jacob", "email": "Jacob@gmail.com"}
    
])

insert_orders=orders.insert().values([
    {"id": 101,"user_id":1 ,"total": 250},
    {"id": 102,"user_id":2 ,"total": 300},
    {"id": 103,"user_id":3 ,"total": 400},
    {"id": 104,"user_id":4 ,"total": 450}
])

conn.execute(insert_users)
conn.execute(insert_orders)
conn.commit()
```

```python
result=conn.execute(text("Select * from users"))
for row in result:
    new_row=list(row)

    print(new_row)
```

```python
result=conn.execute(text("Select * from orders"))
for row in result:
    new_row=list(row)

    print(new_row)
```

```python
join_stmt = select(
        users.c.name,
        users.c.email,
        orders.c.id.label("order_id"),
        orders.c.total,
    ).select_from(users.join(orders, users.c.id == orders.c.user_id))

print("\n--- User Orders ---")
for row in conn.execute(join_stmt):
    print(f"User: {row.name} | Order ID: {row.order_id} | Total: ${row.total}")

    # 3. Left Outer Join (Fetch all users, including those without orders)
outer_join_stmt = select(
    users.c.name,
    orders.c.id.label("order_id"),
    ).select_from(users.outerjoin(orders, users.c.id == orders.c.user_id))

print("\n--- All Users (Including No Orders) ---")
for row in conn.execute(outer_join_stmt):
    print(f"User: {row.name} | Order ID: {row.order_id}")
```
