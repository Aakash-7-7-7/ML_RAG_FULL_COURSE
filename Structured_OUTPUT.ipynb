```python
from pydantic import BaseModel , Field

class Movie(BaseModel):
    title:str=Field(description="The title of the movie")
    year:int=Field(description="The release year of movie")
    rating:str=Field(description="Movie rating out of 10")
    cast:str=Field(description="Cast of the Movie , show 4 main actors")
    director:str=Field(description="Name of Movie Director")
    #summary:str=Field(description="Short Summary of movie")

```

```python
from langchain_openai import ChatOpenAI

model=ChatOpenAI(
    model="qwen/qwen3-vl-4b",
    base_url="http://127.0.0.1:1234/v1",
    api_key="dummy",
    temperature=0.1
)


model_with_structure=model.with_structured_output(Movie)

model_with_structure.invoke("Provide details about movie Ghost Rider")
```

```python

model_with_structure=model.with_structured_output(Movie,include_raw=True)

response=model_with_structure.invoke("Provide details about movie Ghost Rider")
response
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool


@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b


model = ChatOpenAI(
    model="qwen/qwen3-vl-4b",
    base_url="http://127.0.0.1:1234/v1",
    api_key="dummy",
    temperature=0.1
)


agent = create_agent(
    model=model,
    tools=[multiply],

    middleware=[
        SummarizationMiddleware(
            model=model,
            max_tokens=4000
        )
    ]
)


while True:

    question = input("\nYou: ")

    if question.lower() == "exit":
        break

    response = agent.invoke({
        "messages": [
            {
                "role": "user",
                "content": question
            }
        ]
    })

    print("AI:", response["messages"][-1].content)
```
