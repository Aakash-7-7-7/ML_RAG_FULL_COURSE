```python
from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader,UnstructuredPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader=PyMuPDFLoader("PDFS/Artificial_Intelligence_in_Healthcare.pdf")

docs=loader.load()

recursive_splitter=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

pdf_chunks=recursive_splitter.split_documents(docs)

print(f"Created {len(pdf_chunks)} smaller chunks from the PDF")


print(f" First Chunk: {pdf_chunks[0].page_content[:100]}.... ")
```
